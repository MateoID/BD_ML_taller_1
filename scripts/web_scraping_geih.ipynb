```python
import os
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
df_geih = pd.DataFrame()
counter = 0

for i in range(4): 
    driver = webdriver.Chrome()
    
    for n in range(3):
        driver.get(f"https://ignaciomsarmiento.github.io/GEIH2018_sample/page{counter}.html")
        time.sleep(8)
        elemento_web = driver.find_element(By.XPATH, '//table')
        df = pd.read_html(elemento_web.get_attribute('outerHTML'))[0]
        df_i = df.iloc[:, 1:]
        df_geih = pd.concat([df_geih, df_i])
        print(f'Scraping página {counter}.') 
        
        if counter == 10:
          break
        
        counter += 1
    
    if counter == 10:
        print('Finaliza el scraping.')
        break
    
    driver.quit()
```
